```python
import pandas as pd
import os

def extract(fullpath):

    try:
        file_name = os.path.basename(fullpath)      
        df = pd.read_excel(fullpath,engine="openpyxl")
        print(f'{file_name} File extracted succeefully')
        return df                 
                           
    except FileNotFoundError:
        print(f" Error: The file at '{fullpath}' was not found.")
    except Exception as e:
        print(f" An error occurred: {e}")
                           


if __name__ == "__main__":
    File_path = r"C:\Users\ADMIN\Downloads"
    File_name = 'sales_data.xlsx'
    fullpath = os.path.join(File_path,File_name)
    
    df = extract(fullpath)

    if df is not None:
        print("Dataset summary ")
        print(df.info())
    else:
        print("Dataset Empty")
```
